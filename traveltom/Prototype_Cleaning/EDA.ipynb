```python
import pandas as pd
from pathlib import Path
import numpy as np
```

```python
filepath = Path("../datasets/business_SB.parquet")
df = pd.read_parquet(filepath)
```

```python
print("Dataset Shape: ", df.shape)
print("---------------------------")
print(df.head(3))
```

```python
print(df["city"].value_counts().head())
print(df.columns)
```

```python
print(df.info)
print(df.describe)
```

```python
df["popularity"] = df["stars"] * np.log1p(df["review_count"])
```

```python
z = (df["popularity"] - df["popularity"].mean()) / df["popularity"].std()
outliers = df.loc[z.abs() > 3, ["name","popularity","stars","review_count"]]
print(outliers.head(10))
```
